```python
import re

text = "Hello!!! My name is Ganesh123. I scored 95% in exam."

# Remove special characters and numbers
clean_text = re.sub(r'[^a-zA-Z\s]', '', text)

print(clean_text)
```

Alternative Code (Using string)

```python
import string

text = "Hello!!! My name is Ganesh123."

clean_text = ''.join(char for char in text if char.isalpha() or char.isspace())

print(clean_text)
```
