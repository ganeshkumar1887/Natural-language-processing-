re.search()


```python
import re

text = "I love Python"

result = re.search("Python", text)

print(result)
```

re.match()

```python
import re

text = "Python is easy"

result = re.match("Python", text)

print(result)
```

**re.findall():**

```python
import re

text = "cat bat rat"

result = re.findall("at", text)

print(result)
```

**re.sub()**

```python
import re

text = "My age is 22"

clean_text = re.sub(r'\d+', '', text)

print(clean_text)
```

**re.split()**

```python
import re

text = "apple,banana,orange"

result = re.split(",", text)

print(result)
```

Remove URLs

```python
import re

text = "Visit my website https://example.com for more information"

clean_text = re.sub(r'https?://\S+|www\.\S+', '', text)

print(clean_text)
```
